````python
import json

import toonstream
from toonstream import decode, encode

print(f"ToonStream version: {toonstream.__version__}")
````

## 1. Simple Object Encoding

TRON uses `key=value;` syntax instead of JSON's `{"key": "value"}`

````python
# Simple object
data = {
    "name": "Alice Johnson",
    "age": 30,
    "email": "alice@example.com",
    "active": True,
    "score": 95.5
}

# Encode to TRON
tron_str = encode(data, format='tron')
print("TRON output:")
print(tron_str)
print()

# Compare with JSON
json_str = json.dumps(data, separators=(',', ':'))
print(f"JSON Compact: {len(json_str)} chars")
print(f"TRON:         {len(tron_str)} chars")
print(f"Savings:      {(1 - len(tron_str)/len(json_str))*100:.1f}%")
````

## 2. Nested Objects (Dot Notation)

TRON flattens nested objects using dot notation: `user.profile.name=Alice`

````python
# Nested object
data = {
    "user": {
        "name": "Bob",
        "profile": {
            "age": 28,
            "city": "NYC"
        }
    },
    "settings": {
        "theme": "dark",
        "notifications": True
    }
}

tron_str = encode(data, format='tron')
print("TRON output:")
print(tron_str)
print()
print("Notice: Nested keys become dot notation (user.profile.age=28)")
````

## 3. Arrays

TRON arrays use `[item1,item2,item3]` syntax

````python
data = {
    "tags": ["python", "machine-learning", "nlp"],
    "scores": [95, 87, 92, 78],
    "flags": [True, False, True]
}

tron_str = encode(data, format='tron')
print("TRON output:")
print(tron_str)
print()
print("Notice: Booleans become 1/0")
````

## 4. Tabular Data (Arrays of Objects)

TRON uses `@header|row1|row2|row3` format for maximum compression

````python
# Tabular data - employees
employees = [
    {"id": 1, "name": "Alice", "dept": "Engineering", "salary": 95000},
    {"id": 2, "name": "Bob", "dept": "Sales", "salary": 75000},
    {"id": 3, "name": "Carol", "dept": "Marketing", "salary": 85000},
    {"id": 4, "name": "David", "dept": "Engineering", "salary": 105000},
    {"id": 5, "name": "Eve", "dept": "HR", "salary": 65000},
]

tron_str = encode(employees, format='tron')
json_str = json.dumps(employees, separators=(',', ':'))

print("TRON output:")
print(tron_str)
print()
print(f"JSON Compact: {len(json_str)} chars")
print(f"TRON:         {len(tron_str)} chars")
print(f"Savings:      {(1 - len(tron_str)/len(json_str))*100:.1f}%")
````

## 5. Special Values

| Python | TRON |
|--------|------|
| `True` | `1` |
| `False` | `0` |
| `None` | `~` |

````python
data = {
    "enabled": True,
    "disabled": False,
    "empty": None,
    "count": 0,
    "pi": 3.14159
}

tron_str = encode(data, format='tron')
print("TRON output:")
print(tron_str)
````

## 6. Decoding TRON

Use `decode(tron_str, format='tron')` to parse TRON back to Python

````python
# Decoding examples
tron_samples = [
    'name=Alice;age=30;active=1',
    'user.name=Bob;user.age=25',
    '@id,name|1,Alice|2,Bob|3,Carol',
]

for tron in tron_samples:
    decoded = decode(tron, format='tron')
    print(f"TRON: {tron}")
    print(f"Data: {decoded}")
    print()
````

## 7. Format Comparison: JSON vs TOON vs TRON

````python
# Real-world API response
api_response = {
    "status": "success",
    "data": {
        "users": [
            {"id": 1, "name": "Alice", "email": "alice@example.com", "role": "admin"},
            {"id": 2, "name": "Bob", "email": "bob@example.com", "role": "user"},
            {"id": 3, "name": "Carol", "email": "carol@example.com", "role": "user"},
        ],
        "total": 3,
        "page": 1
    },
    "meta": {
        "version": "2.0.0",
        "timestamp": "2025-12-01T12:00:00Z"
    }
}

json_pretty = json.dumps(api_response, indent=2)
json_compact = json.dumps(api_response, separators=(',', ':'))
toon_str = encode(api_response, format='toon')
tron_str = encode(api_response, format='tron')

print("Format Comparison:")
print(f"  JSON (pretty):   {len(json_pretty):>4} chars")
print(f"  JSON Compact:    {len(json_compact):>4} chars")
print(f"  TOON:            {len(toon_str):>4} chars")
print(f"  TRON:            {len(tron_str):>4} chars")
print()
print(f"TRON saves {(1 - len(tron_str)/len(json_pretty))*100:.1f}% vs JSON Pretty")
print(f"TRON saves {(1 - len(tron_str)/len(json_compact))*100:.1f}% vs JSON Compact")
````

## 8. Round-Trip Verification

````python
# Verify lossless round-trip
original = {
    "name": "Test User",
    "scores": [100, 95, 87],
    "settings": {
        "dark_mode": True,
        "volume": 0.8
    }
}

# Encode
tron_str = encode(original, format='tron')
print(f"Encoded: {tron_str}")

# Decode
decoded = decode(tron_str, format='tron')
print(f"Decoded: {decoded}")

# Verify
print(f"\nRound-trip match: {original == decoded}")
````

## TRON Syntax Summary

| Element | TRON Syntax | Example |
|---------|-------------|--------|
| Key-Value | `key=value;` | `name=Alice;age=30` |
| Nested | `parent.child=value` | `user.profile.name=Alice` |
| Array | `[item1,item2]` | `tags=[python,ml,ai]` |
| Tabular | `@col1,col2\|row1\|row2` | `@id,name\|1,Alice\|2,Bob` |
| True | `1` | `active=1` |
| False | `0` | `enabled=0` |
| Null | `~` | `value=~` |
| String (special) | `"quoted"` | `msg="hello;world"` |

## When to Use TRON

✅ **Best for:**
- LLM context optimization
- Simple key-value configs
- Nested configurations
- Streaming data to LLMs
- Logging and telemetry

## API Reference

```python
# Encode to TRON
tron_str = encode(data, format='tron')

# Decode from TRON
data = decode(tron_str, format='tron')

# Encode to TOON
toon_str = encode(data, format='toon')

# Decode from TOON
data = decode(toon_str, format='toon')
```
